Q5. Date Format Consistency

The Order_Date column has multiple formats.

a) List all date formats present in the dataset

b) Why is this a problem during data loading?

Sol :    

a) The date formats present are DD-MM-YYYY (e.g., 12-01-2024) and YYYY/MM/DD (e.g., 2024/01/18).

b) This is a problem because databases typically expect a single, standardized date format for a column. Mixed formats can lead to parsing errors, records being rejected during the load, or dates being stored incorrectly (e.g., swapping months and days).

Q7. Pre-Load Validation Checklist

List the exact pre-load validation checks you would perform on this dataset before loading

Sol :     
- Uniqueness/Deduplication Check: Ensure the Primary Key (Order_ID) contains no duplicates.
- Data Type Check: Verify that Sales_Amount contains only numeric values.
- Null/Completeness Check: Identify and flag any missing or "Null" values in mandatory columns.
- Format Consistency Check: Ensure all dates in Order_Date follow a single standard format.

Q8. Cleaning Strategy

Describe the step-by-step cleaning actions required to make this dataset load-ready

Sol :         

- Deduplicate: Remove the duplicate row for Order_ID O101, keeping only one instance.
- Fix Data Types: Convert the string "Three Thousand" to the numeric value 3000 in the Sales_Amount column.
- Standardize Formats: Convert the date 2024/01/18 to match the rest of the dataset's format (e.g., 18-01-2024), or convert all dates to a standard SQL format like YYYY-MM-DD.
- Handle Missing Values: Address the "Null" in O102 by either imputing a default value (like 0), estimating it, or discarding the row based on business rules.

Q9. Loading Strategy Selection

Assume this dataset represents daily sales data.

a) Should a Full Load or Incremental Load be used?

b) Justify your choice

Sol :       
a) An Incremental Load should be used.

b) Justification: Daily sales data represents new transactions occurring over time. It is highly inefficient to wipe the entire historical table and reload everything (Full Load) just to add one new day of data. An Incremental Load will only append the new daily records, saving significant time and computational resources.

Q10. BI Impact Scenario

Assume this dataset was loaded without cleaning and connected to a BI dashboard.

a) What incorrect results might appear in Total Sales KPI?

b) Which records specifically would cause misleading insights?

c) Why would BI tools not detect these issues automatically?

Sol :        
a) The Total Sales KPI would be artificially inflated due to the duplicate record being counted twice. Additionally, depending on how the BI tool handles errors, the row with "Three Thousand" might be omitted entirely, underreporting the actual sales.

b) Records O101 (duplicate), O102 (Null/Missing), and O104 (Text instead of number) would cause misleading insights.

c) BI tools are generally designed to read, aggregate, and visualize data "as-is." They assume that the data in the warehouse has already been cleaned and validated by the ETL pipeline. They do not inherently know that a repeated ID is an error rather than two distinct sales, nor do they automatically fix bad source data.
